## OOP and Custom Exception

```python
# إنشاء Custom Exception وBase Class
# إنشاء استثناء مخصص للسعر غير الصحيح
class InvalidPriceError(Exception):
    pass


# إنشاء الفئة الأساسية
class House:

    def __init__(self, price, bedrooms, city):

        # التحقق من صحة السعر
        if price <= 0:
            raise InvalidPriceError("The price must be greater than zero.")

        self.price = price
        self.bedrooms = bedrooms
        self.city = city

    # دالة لعرض معلومات المنزل
    def display_info(self):
        print("City", self.city)
        print("Priec:", self.price)
        print(" Bedrooms:", self.bedrooms)
```

```python
# انشاء فئتين فرعيتين
# الفئة الفرعية الأولى
class LuxuryHouse(House):

    def house_type(self):
        return "Luxury house"


# الفئة الفرعية الثانية
class NormalHouse(House):

    def house_type(self):
        return "Ordinary house"
```

```python
# من الصفوف 10 الى 19 object   انشاء
# إنشاء قائمة لتخزين الكائنات
houses = []

# استخدام الصفوف من 10 إلى 19
for i in range(10, 20):

    row = df.iloc[i]

    try:
        # إذا كان السعر 500000 أو أكثر نعتبر المنزل فاخرًا
        if row["price"] >= 500000:

            house = LuxuryHouse(
                row["price"],
                row["bedrooms"],
                row["city"]
            )

        else:

            house = NormalHouse(
                row["price"],
                row["bedrooms"],
                row["city"]
            )

        # إضافة الكائن إلى القائمة
        houses.append(house)

    except InvalidPriceError as error:
        print("Error in class", i, ":", error)
```

```python
# عرض معلومات المنازل
for house in houses:

    print("Type of house", house.house_type())

    house.display_info()

    print("--------------------")
```

# Milestone 4: Pandas Analysis and Excel Export


## Data Cleaning


```python
# إنشاء نسخة من البيانات الأصلية
df_clean = df.copy()

print("Dataset shape before cleaning:", df_clean.shape)
```

```python
# تنظيف أسماء الأعمدة
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df_clean.columns)
```

```python
# فحص القيم الناقصة
print("Missing values before cleaning:")
print(df_clean.isnull().sum())

# حذف الصفوف التي تحتوي على قيم ناقصة
df_clean = df_clean.dropna()

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# فحص وحذف الصفوف المكررة
duplicates_before = df_clean.duplicated().sum()
print("Duplicate rows before cleaning:", duplicates_before)

df_clean = df_clean.drop_duplicates()

duplicates_after = df_clean.duplicated().sum()
print("Duplicate rows after cleaning:", duplicates_after)
```

```python
# تحويل عمود التاريخ إلى Date Type
df_clean["date"] = pd.to_datetime(df_clean["date"])

print("Date column type:", df_clean["date"].dtype)
print("Dataset shape after cleaning:", df_clean.shape)
```

# Analysis Questions


## Question 1

**Which city has the highest average house price?**


```python
# حساب متوسط سعر المنازل في كل مدينة
average_price_by_city = (
    df_clean.groupby("city")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(average_price_by_city.head())

top_city = average_price_by_city.index[0]
top_city_price = average_price_by_city.iloc[0]

print("\nAnswer:")
print(top_city, "has the highest average house price:", top_city_price)
```

## Question 2

**What is the average house price for each condition value?**


```python
# حساب متوسط السعر حسب حالة المنزل
average_price_by_condition = (
    df_clean.groupby("condition")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(average_price_by_condition)

best_condition = average_price_by_condition.index[0]
best_condition_price = average_price_by_condition.iloc[0]

print("\nAnswer:")
print("Condition", best_condition, "has the highest average price:", best_condition_price)
```

## Question 3

**Which city has the largest number of houses?**


```python
# حساب عدد المنازل في كل مدينة
houses_by_city = (
    df_clean.groupby("city")
    .size()
    .sort_values(ascending=False)
)

print(houses_by_city.head())

city_with_most_houses = houses_by_city.index[0]
most_houses_count = houses_by_city.iloc[0]

print("\nAnswer:")
print(city_with_most_houses, "has the largest number of houses:", most_houses_count)
```

# Analysis Summary


```python
# إنشاء ملخص للأسئلة والإجابات
analysis_summary = pd.DataFrame({
    "Question": [
        "Which city has the highest average house price?",
        "What condition has the highest average house price?",
        "Which city has the largest number of houses?"
    ],
    "Answer": [
        top_city + " - " + str(round(top_city_price, 2)),
        "Condition " + str(best_condition) + " - " + str(round(best_condition_price, 2)),
        city_with_most_houses + " - " + str(most_houses_count) + " houses"
    ]
})

analysis_summary
```

# Pivot Table


```python
# Pivot Table باستخدام أعمدة تصنيفية ورقمية
pivot_table = pd.pivot_table(
    df_clean,
    index="condition",
    columns="waterfront",
    values="price",
    aggfunc="mean"
)

pivot_table
```

# Excel Export


```python
# تصدير البيانات النظيفة وPivot Table وملخص التحليل إلى Excel
report_file = "Houses_Prices_Report.xlsx"

with pd.ExcelWriter(report_file, engine="openpyxl") as writer:
    df_clean.to_excel(
        writer,
        sheet_name="Cleaned Data",
        index=False
    )

    pivot_table.to_excel(
        writer,
        sheet_name="Pivot Table"
    )

    analysis_summary.to_excel(
        writer,
        sheet_name="Analysis Summary",
        index=False
    )

print("Excel file created successfully:", report_file)
```
